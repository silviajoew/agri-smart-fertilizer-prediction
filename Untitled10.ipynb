```python
import pandas as pd
import pickle
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score

# 1. LOAD & CLEAN
df = pd.read_csv('fertilizer_recommendation.csv')
df.columns = df.columns.str.strip().str.replace(' ', '_').str.lower()

# 2. ADVANCED FEATURE ENGINEERING
encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le

target = 'recommended_fertilizer'
X = df.drop(target, axis=1)
y = df[target]

# Feature Scaling - Vital for Gradient Boosting
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.15, random_state=42)

# 3. XGBOOST MODEL (The Accuracy King)
# We use more estimators and a lower learning rate for precision
model = XGBClassifier(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=8,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    use_label_encoder=False,
    eval_metric='mlogloss'
)

print("Training the High-Accuracy Booster...")
model.fit(X_train, y_train)

# 4. EVALUATE
y_pred = model.predict(X_test)
final_acc = accuracy_score(y_test, y_pred)
print(f"✅ UPGRADED ACCURACY: {final_acc * 100:.2f}%")

# 5. SAVE THE EXPERT BRAIN
project_brain = {
    'model': model,
    'scaler': scaler,
    'encoders': encoders,
    'features': X.columns.tolist(),
    'rec_map': {
        'Urea': 'Critical Nitrogen shortage. Apply Urea for leaf greening.',
        'DAP': 'Phosphorus needed for root establishment.',
        'MOP': 'Potassium required for crop immunity.',
        '14-35-14': 'Multi-nutrient support for early growth.'
    }
}

with open('fertilizer_expert_system.pkl', 'wb') as f:
    pickle.dump(project_brain, f)
```

```python

import pandas as pd
import pickle
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score

# 2. LOAD & CLEAN
df = pd.read_csv('fertilizer_recommendation.csv')
df.columns = df.columns.str.strip().str.replace(' ', '_').str.lower()

# 3. ENCODING & SCALING
encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le

X = df.drop('recommended_fertilizer', axis=1)
y = df['recommended_fertilizer']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. STRATIFIED SPLIT
# This ensures that rare fertilizers are represented equally in training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

# 5. THE ACCURACY BOOSTER (GridSearch)
# We define a search space for the model to find the best 'brain'
xgb_model = XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='mlogloss')

param_grid = {
    'n_estimators': [500, 700],
    'max_depth': [6, 10, 15],
    'learning_rate': [0.01, 0.05, 0.1],
    'gamma': [0, 0.1, 0.2]
}

print("Running Grid Search to hit 95% Accuracy... this may take 2-3 minutes.")
grid_search = GridSearchCV(xgb_model, param_grid, cv=3, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)

# 6. RESULTS
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)
print(f"✅ NEW TARGET ACCURACY: {accuracy_score(y_test, y_pred) * 100:.2f}%")

# 7. EXPORT
project_brain = {
    'model': best_model,
    'scaler': scaler,
    'encoders': encoders,
    'features': X.columns.tolist(),
    'rec_map': {
        'Urea': 'Apply Urea: High Nitrogen required for current crop stage.',
        'DAP': 'Apply DAP: Phosphorus boost needed for root establishment.',
        'MOP': 'Apply MOP: Potassium required for fruit quality and stress resistance.',
        '14-35-14': 'Apply 14-35-14: Balanced nutrient profile recommended.'
    }
}

with open('fertilizer_expert_system.pkl', 'wb') as f:
    pickle.dump(project_brain, f)
```

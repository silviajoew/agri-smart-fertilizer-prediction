```python
import pandas as pd
import pickle
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score

# 1. LOAD & CLEAN DATA
# Ensure 'fertilizer_recommendation.csv' is in your Fertilizer_Project folder
df = pd.read_csv('fertilizer_recommendation.csv')
df.columns = df.columns.str.strip().str.replace(' ', '_').str.lower()

# 2. FEATURE ENCODING
encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le

X = df.drop('recommended_fertilizer', axis=1)
y = df['recommended_fertilizer']

# 3. FEATURE SCALING
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. STRATIFIED SPLIT
# This ensures all fertilizer types are represented equally for better accuracy
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

# 5. OPTIMIZED XGBOOST MODEL
# Parameters tuned to maximize accuracy on 10,000 records
model = XGBClassifier(
    n_estimators=1000,       # More trees to capture complex patterns
    learning_rate=0.01,      # Lower rate for higher precision
    max_depth=12,            # Deeper trees to catch specific soil features
    gamma=0.2,               # Prevents overfitting (memorization)
    subsample=0.9,
    colsample_bytree=0.9,
    random_state=42,
    eval_metric='mlogloss'
)

print("Training high-precision model... Please wait.")
model.fit(X_train, y_train)

# 6. EVALUATE ACCURACY
y_pred = model.predict(X_test)
final_acc = accuracy_score(y_test, y_pred)
print(f"✅ FINAL ACCURACY: {final_acc * 100:.2f}%")

# 7. EXPORT THE 'BRAIN'
# This saves the model directly into your project folder
project_brain = {
    'model': model,
    'scaler': scaler,
    'encoders': encoders,
    'features': X.columns.tolist(),
    'rec_map': {
        'Urea': 'Apply Urea: High Nitrogen required for current crop stage.',
        'DAP': 'Apply DAP: Phosphorus boost needed for root establishment.',
        'MOP': 'Apply MOP: Potassium required for fruit quality.',
        '14-35-14': 'Apply 14-35-14: Balanced nutrient profile recommended.'
    }
}

with open('fertilizer_expert_system.pkl', 'wb') as f:
    pickle.dump(project_brain, f)
    print("Success! 'fertilizer_expert_system.pkl' has been updated.")
```

```python
# Install the balancing library if you haven't already
!pip install imbalanced-learn xgboost
import pandas as pd
import pickle
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
from imblearn.over_sampling import SMOTE 

# 1. LOAD & CLEAN
df = pd.read_csv('fertilizer_recommendation.csv')
df.columns = df.columns.str.strip().str.replace(' ', '_').str.lower()

# 2. ENCODING
encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le

X = df.drop('recommended_fertilizer', axis=1)
y = df['recommended_fertilizer']

# 3. ADVANCED SCALING
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. DATA BALANCING (The Secret to 95%+)
# SMOTE creates synthetic examples of rare fertilizer classes so the model learns better
smote = SMOTE(random_state=42)
X_balanced, y_balanced = smote.fit_resample(X_scaled, y)

X_train, X_test, y_train, y_test = train_test_split(
    X_balanced, y_balanced, test_size=0.2, random_state=42, stratify=y_balanced
)

# 5. ULTRA-OPTIMIZED XGBOOST
model = XGBClassifier(
    n_estimators=1200,      # Increased trees for deeper learning
    learning_rate=0.01,     # Very small steps for maximum precision
    max_depth=15,           # Deeper trees to capture subtle soil differences
    min_child_weight=1,
    gamma=0.4,              # Increased penalty to prevent overfitting
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric='mlogloss'
)

print("Training ultra-high precision model... this may take 1-2 minutes.")
model.fit(X_train, y_train)

# 6. EVALUATE
y_pred = model.predict(X_test)
final_acc = accuracy_score(y_test, y_pred)
print(f"✅ NEW TARGET ACCURACY: {final_acc * 100:.2f}%")

# 7. EXPORT
project_brain = {
    'model': model,
    'scaler': scaler,
    'encoders': encoders,
    'features': X.columns.tolist(),
    'rec_map': {
        'Urea': 'Apply Urea: High Nitrogen required for current crop stage.',
        'DAP': 'Apply DAP: Phosphorus boost needed for root establishment.',
        'MOP': 'Apply MOP: Potassium required for fruit quality.',
        '14-35-14': 'Apply 14-35-14: Balanced nutrient profile recommended.'
    }
}

with open('fertilizer_expert_system.pkl', 'wb') as f:
    pickle.dump(project_brain, f)
```
